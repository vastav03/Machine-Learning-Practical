```python
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Study_Hours": [2, 4, 6, 8, 10],
    "Marks": [45, 55, 68, 78, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())      # First five rows
print(df.shape)        # Number of rows and columns
print(df.columns)      # Column names
print(df.info())       # Data types and information
print(df.describe())   # Statistical summary
```

```python
print(df["Study_Hours"])

print(df[["Study_Hours", "Marks"]])
```

```python
print("Average study hours:", df["Study_Hours"].mean())
print("Average marks:", df["Marks"].mean())
print("Maximum marks:", df["Marks"].max())
print("Minimum marks:", df["Marks"].min())
```

```python
import matplotlib.pyplot as plt

study_hours = [2, 4, 6, 8, 10]
marks = [45, 55, 68, 78, 90]

plt.plot(study_hours, marks, marker='o')

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
marks = [45, 55, 68, 78, 90, 62, 74, 81, 56, 69]

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()

# Convert to Pandas DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["Target"] = iris.target

# Display dataset
print("First five records:")
print(df.head())

# Dataset information
print("\nDataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
print(df.describe())

# NumPy operation
print("\nMean of features:")
print(np.mean(iris.data, axis=0))

# Visualization
plt.scatter(
    df["sepal length (cm)"],
    df["petal length (cm)"]
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.title("Iris Dataset")

plt.show()
```
